```python
import pandas as pd
import glob
import os
```

```python
def validate_data_integrity(df: pd.DataFrame, df_all_flights: pd.DataFrame) -> None:
    """
    Validate the integrity of the combined data.
    """
    if set(df['flight_id']) != set(df_all_flights['flight_id']):
        combined_set = set(df['flight_id'])
        all_flights_set = set(df_all_flights['flight_id'])

        missing_in_combined = all_flights_set - combined_set
        missing_in_all_flights = combined_set - all_flights_set

        print(f"Missing flights in combined set: {missing_in_combined}")
        print(f"Missing flights in all flights set: {missing_in_all_flights}")

def find_excel_files(directory):
    # Use glob to find all Excel files (both .xlsx and .xls) in the directory and subdirectories
    excel_files = glob.glob(os.path.join(directory, '**', '*.xlsx'), recursive=True)
    excel_files.extend(glob.glob(os.path.join(directory, '**', '*.xls'), recursive=True))

    return excel_files
```

```python
df_all_flights = pd.read_excel('../../data/interim/flights/all_flights.xlsx')
set_of_flights = set(df_all_flights['flight_id'])

directory_path = '../../data/raw/api/bereik/'
excel_files = find_excel_files(directory_path)
```

```python
for file in excel_files:
    df = pd.read_excel(file, dtype={'start_date': str, 'end_date': str})
    validate_data_integrity(df, df_all_flights)
```

```python
import pandas as pd
import numpy as np
import math
import hashlib
import uuid
from datetime import timedelta
from tqdm import tqdm


def generate_uuid_from_columns(row):
    """
    Generates a (unique) UUID based on the values of the specified columns in the given row.
    Args:
        row (dict): A dictionary representing a row of data.
    Returns:
        uuid.UUID or None: The generated UUID or None if the UUID format is invalid.
    """
    # Concatenate values of the specified columns into a single string
    # Combination of advertiser, product, and start_date; no 'end-date' because the campaign can still run at this moment

    if row['mock_campaign']:

        # For mock campaigns, include start_date and end_date in the UUID generation
        concatenated_string = ''.join(
            str('mock_' + row[col]) for col in ['advertiser', 'product', 'start_date', 'end_date']).lower().replace(" ", "")
    else:

        # For original campaigns, use the existing logic
        concatenated_string = ''.join(
            str(row[col]) for col in ['advertiser', 'product', 'start_date']).lower().replace(" ", "")

    # Hash the concatenated string using sha256
    hashed_value = hashlib.sha256(concatenated_string.encode()).digest()

    # Convert hash to UUID
    try:
        new_uuid = str(uuid.UUID(bytes=hashed_value[:16]))
        return new_uuid
    except ValueError:
        print("Invalid UUID format:", hashed_value)
        return None
```

```python
df_mocks = pd.read_excel('../../data/interim/flights/all_mock_campaigns.xlsx')
df_mocks['start_date'] = pd.to_datetime(df_mocks['start_date']).dt.strftime("%d%m%y")
df_mocks['end_date'] = pd.to_datetime(df_mocks['end_date']).dt.strftime("%d%m%y")
df_mocks['mock_campaign'] = True
```

```python
df_mocks['flight_id'] = df_mocks.apply(generate_uuid_from_columns, axis=1)
```
