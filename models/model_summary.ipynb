```python
import os
import json
import joblib
import pandas as pd
import numpy as np
```

```python
results_file = "model_results.json"
def load_results():
    """Load results from JSON file if it exists."""
    if os.path.exists(results_file):
        with open(results_file, "r") as f:
            return json.load(f)
    return {}
```

```python
data = load_results()
rows = []
for target_group, models in data.items():
    for model, dates in models.items():
        for date, content in dates.items():
            row = {
                "target_group": target_group,
                "model": model,
                "date": date,
                **{f"train_{k}": v for k, v in content["metrics"]["train"].items()},
                **{f"test_{k}": v for k, v in content["metrics"]["test"].items()},
                "best_params": content["best_params"],
            }
            rows.append(row)

# Create the DataFrame
df = pd.DataFrame(rows)
df
```
